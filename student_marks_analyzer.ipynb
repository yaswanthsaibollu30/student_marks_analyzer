```python
students,marks=[],[]
```

```python
for i in range(5):
    name= input("enter name")
    st_marks=int(input("enter marks"))
    students.append(name)
    marks.append(st_marks)
```

```python
print(students)
print(marks)
```

```python
avg = sum(marks)/len(marks)
```

```python
highest_marks = max(marks)
lowest_marks = min(marks)
```

```python
print(f"the average marks = {avg}")
print("highest marks =",highest_marks)
print("lowest marks = ",lowest_marks)
```

```python
grade =[]
```

```python
for i in marks:
    if i>=0 and i <=100:
        if i >=95:
            st_grade = "A+"
            grade.append(st_grade)
        elif i>=90:
            st_grade = "A"
            grade.append(st_grade)
        elif i>=75:
            st_grade ="B"
            grade.append(st_grade)
        elif i>=65:
            st_grade = "C"
            grade.append(st_grade)
        elif i>=45:
            st_grade = "D"
            grade.append(st_grade)
        elif i>=35:
            st_grade = "E"
            grade.append(st_grade)
        else:
            st_grade = "Fail"
            grade.append(st_grade)
    else:
        print("invalid marks")
```

```python
grade
```

```python
print("students = ",students)
print("Marks = ",marks)
print("Grade = ",grade)
```

```python
for name,marks,gra in zip(students,marks,grade):
    print(f"Name : {name} | Marks : {marks} | Grade : {gra}")
```
